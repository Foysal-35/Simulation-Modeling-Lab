# Necessary Libraries

```python
import numpy as np
from sympy import Matrix
import matplotlib.pyplot as plt
import seaborn as sns
```

# Part 1 — Data Generation


# Generate Uniform Distribution Data


```python
uniform_wait = np.random.uniform(2, 12, 3000)

print("First 10 Uniform Wait Times:")
print(uniform_wait[:10])
```

# Generate Normal Distribution Data

```python
normal_wait = np.random.normal(8, 2, 3000)

# Replace negative values with 0
normal_wait = np.maximum(normal_wait, 0)

print("First 10 Normal Wait Times:")
print(normal_wait[:10])
```

# Part 2 — Visualization

# Histogram for Uniform Distribution

```python
plt.figure(figsize=(6,4))

plt.hist(uniform_wait, bins=30, edgecolor='black')

plt.title("Uniform Distribution")
plt.xlabel("Wait Time (minutes)")
plt.ylabel("Frequency")

plt.show()
```

# Histogram for Normal Distribution

```python
plt.figure(figsize=(6,4))

plt.hist(normal_wait, bins=30, edgecolor='black')

plt.title("Normal Distribution")
plt.xlabel("Wait Time (minutes)")
plt.ylabel("Frequency")

plt.show()
```

# Two Histograms in Same Figure

```python
plt.figure(figsize=(12,5))

plt.subplot(1,2,1)
plt.hist(uniform_wait, bins=30, edgecolor='black')
plt.title("Uniform Distribution")

plt.subplot(1,2,2)
plt.hist(normal_wait, bins=30, edgecolor='black')
plt.title("Normal Distribution")

plt.tight_layout()
plt.show()
```

# Part 3 — Statistical Analysis

# Statistical Analysis for Uniform Distribution

```python
print("Uniform Distribution Statistics")

print("Mean =", np.mean(uniform_wait))
print("Median =", np.median(uniform_wait))
print("Standard Deviation =", np.std(uniform_wait))
print("Maximum Wait Time =", np.max(uniform_wait))
```

# Statistical Analysis for Normal Distribution

```python
print("Normal Distribution Statistics")

print("Mean =", np.mean(normal_wait))
print("Median =", np.median(normal_wait))
print("Standard Deviation =", np.std(normal_wait))
print("Maximum Wait Time =", np.max(normal_wait))
```
